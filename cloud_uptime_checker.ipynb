# Project 6: Cloud Infrastructure Uptime & Service Monitor

### Project Overview
This automation script simulates a continuous monitoring agent that audits the availability and response states of enterprise cloud services (such as API endpoints, databases, and compute nodes). It programmatically evaluates system latency, detects service disruptions, and logs critical uptime metrics to mimic corporate automated monitoring tools.

### Skills Demonstrated
* Real-time loop processing and data manipulation in Python
* Systems reliability, monitoring, and infrastructure health tracking
* Automated operational status reporting


```python
import random # Simulated random variable to mimic live network fluctuations

# Simulated Corporate Cloud Services Registry
cloud_services = [
    {"service_name": "Auth-API-Service", "endpoint_type": "API", "expected_status": 200, "simulated_latency_ms": 45},
    {"service_name": "Customer-DB-Cluster", "endpoint_type": "Database", "expected_status": 200, "simulated_latency_ms": 110},
    {"service_name": "Public-Web-Frontend", "endpoint_type": "Web", "expected_status": 200, "simulated_latency_ms": 25},
    {"service_name": "Payment-Gateway-Link", "endpoint_type": "API", "expected_status": 503, "simulated_latency_ms": 0}, # 503 = Service Unavailable
    {"service_name": "Log-Storage-Archive", "endpoint_type": "Storage", "expected_status": 200, "simulated_latency_ms": 320}
]

print("📊 --- STARTING CLOUD INFRASTRUCTURE HEALTH MONITOR --- 📊\n")

operational_count = 0
outage_count = 0

# Loop through services to check availability and performance metrics
for service in cloud_services:
    print(f"Pinging Service Host: {service['service_name']} ({service['endpoint_type']})")

    # SYSTEM EVALUATION RULE 1: Detect Outages based on HTTP Status Codes
    if service['expected_status'] != 200:
        print(f"  ❌ ALERT: SERVICE DISRUPTION DETECTED! Host responded with status code {service['expected_status']}.")
        outage_count += 1
        print("-" * 65)
        continue # Move to the next service immediately

    # SYSTEM EVALUATION RULE 2: Detect Performance Degradation (Latency Check)
    if service['simulated_latency_ms'] > 200:
        print(f"  ⚠️ PERFORMANCE WARNING: High latency detected ({service['simulated_latency_ms']}ms). System is lagging.")
        operational_count += 1
    else:
        print(f"  ✅ HEALTHY: Response received in {service['simulated_latency_ms']}ms. Service is nominal.")
        operational_count += 1

    print("-" * 65)

# Generate Infrastructure Health Statistics
total_services = len(cloud_services)
uptime_percentage = (operational_count / total_services) * 100

print("\n📈 --- SYSTEM PERFORMANCE METRICS ---")
print(f"Total Monitored Hosts: {total_services}")
print(f"Operational States   : {operational_count} Active")
print(f"Critical Failures    : {outage_count} Outage(s)")
print(f"Global Fleet Uptime  : {uptime_percentage:.1f}%")
print("-" * 35)

if outage_count > 0:
    print("OPERATIONS STATUS: ACTION REQUIRED | Incident response triggered. Check service dependencies.")
else:
    print("OPERATIONS STATUS: STABLE | All monitored endpoints meet baseline SLAs.")
```
