# Project 2: Azure Cloud Identity & Access Management (IAM) Auditor

### Project Overview
This automated tool simulates scanning a corporate Microsoft Entra ID (formerly Azure Active Directory)
user database. It programmatically identifies critical security risks, configuration gaps, and compliance
failures—specifically targeting privileged accounts lacking Multi-Factor Authentication (MFA) and stale user passwords.

### Skills Demonstrated
* Data parsing and dictionary auditing in Python
* Access control and identity governance principles
* Cloud security compliance reporting


```python
# Simulated Azure Active Directory User Database
azure_user_directory = [
    {"username": "sara.admin@company.com", "role": "Global Administrator", "mfa_enabled": True, "days_since_password_change": 12},
    {"username": "john.dev@company.com", "role": "Contributor", "mfa_enabled": True, "days_since_password_change": 45},
    {"username": "marketing.guest@company.com", "role": "Reader", "mfa_enabled": False, "days_since_password_change": 120},
    {"username": "alex.engineer@company.com", "role": "User Access Administrator", "mfa_enabled": False, "days_since_password_change": 15},
    {"username": "billing.temp@company.com", "role": "Billing Reader", "mfa_enabled": True, "days_since_password_change": 95}
]

print("🔒 --- STARTING AUTOMATED AZURE IAM SECURITY AUDIT --- 🔒\n")

critical_alerts_found = 0
compliance_warnings_found = 0

# Loop through each user profile to audit their security posture
for user in azure_user_directory:
    print(f"Auditing Account: {user['username']}")
    print(f"Assigned Azure Role: {user['role']}")

    # RISK CHECK 1: Missing MFA on High-Privilege Roles
    if not user['mfa_enabled']:
        if "Administrator" in user['role'] or "Contributor" in user['role']:
            print(f"  🚨 CRITICAL SECURITY RISK: High-privilege role lacks Multi-Factor Authentication!")
            critical_alerts_found += 1
        else:
            print(f"  ⚠️ WARNING: Standard account lacks MFA. Recommend enabling.")
            compliance_warnings_found += 1

    # RISK CHECK 2: Stale Passwords (Over 90 days)
    if user['days_since_password_change'] > 90:
        print(f"  ⚠️ COMPLIANCE ISSUE: Password is stale ({user['days_since_password_change']} days old).")
        compliance_warnings_found += 1

    print("-" * 50)

# Final Audit Summary Report
print("\n📋 --- AUDIT SUMMARY REPORT ---")
print(f"Total Critical Risks: {critical_alerts_found}")
print(f"Total Compliance Warnings: {compliance_warnings_found}")

if critical_alerts_found > 0:
    print(f"STATUS: FAILED | Immediate remediation required for privileged identities.")
else:
    print("STATUS: PASSED | No high-priority identity risks identified.")
```
