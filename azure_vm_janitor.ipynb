# Project 3: Automated Cloud Virtual Machine (VM) Janitor

### Project Overview
This cloud governance script automates infrastructure resource lifecycle management. It scans an enterprise fleet of virtual machines to detect optimization opportunities—specifically targeting abandoned test servers left idling over weekends and unutilized legacy instances. The tool flags underutilized assets to reduce corporate cloud waste.

### Skills Demonstrated
* Conditional multi-criteria evaluation in Python loops
* Cloud resource lifecycle and governance management concepts
* Automated systems maintenance and cost optimization


```python
# Simulated Enterprise Virtual Machine Inventory
azure_vm_inventory = [
    {"vm_name": "vm-prod-web-01", "environment": "Production", "status": "Running", "cpu_utilization_pct": 78, "is_temporary": False},
    {"vm_name": "vm-dev-sandbox-04", "environment": "Development", "status": "Running", "cpu_utilization_pct": 2, "is_temporary": True},
    {"vm_name": "vm-prod-db-02", "environment": "Production", "status": "Running", "cpu_utilization_pct": 64, "is_temporary": False},
    {"vm_name": "vm-test-api-11", "environment": "Testing", "status": "Running", "cpu_utilization_pct": 1, "is_temporary": True},
    {"vm_name": "vm-legacy-reporting", "environment": "Operations", "status": "Stopped", "cpu_utilization_pct": 0, "is_temporary": False}
]

print("🧹 --- STARTING AUTOMATED VIRTUAL MACHINE LIFECYCLE AUDIT --- 🧹\n")

vms_to_shutdown = 0
vms_to_keep_running = 0

# Scan the cloud inventory to identify waste
for vm in azure_vm_inventory:
    print(f"Evaluating Infrastructure: {vm['vm_name']} [{vm['environment']}]")

    # JANITOR RULE 1: Flag running non-prod temporary VMs with ultra-low utilization
    if vm['status'] == "Running" and vm['is_temporary']:
        if vm['cpu_utilization_pct'] < 5:
            print(f"  🚨 JANITOR ACTION: Flagged for DELETION. Idle temporary test server detected (CPU: {vm['cpu_utilization_pct']}%).")
            vms_to_shutdown += 1
            continue  # Skip to the next server since we already flagged this one

    # JANITOR RULE 2: Flag production or operational infrastructure that is completely idle
    if vm['status'] == "Running" and vm['cpu_utilization_pct'] == 0:
        print(f"  ⚠️ WARNING: Server is running but shows 0% utilization. Recommend investigating workload.")
        vms_to_keep_running += 1
    else:
        # If it doesn't meet the cleanup criteria, it is safe
        if vm['status'] == "Running":
            print(f"  ✅ SAFE: Active workload verified (CPU: {vm['cpu_utilization_pct']}%). Keeping operational.")
            vms_to_keep_running += 1
        else:
            print(f"  💤 DORMANT: Server is currently stopped. No active runtime costs accumulating.")

    print("-" * 65)

# Generate Summary Analytics
print("\n📊 --- OPERATIONS CLEANUP SUMMARY ---")
print(f"Infrastructure Retained: {vms_to_keep_running} active servers")
print(f"Infrastructure Flagged for Cleanup: {vms_to_shutdown} abandoned servers")

if vms_to_shutdown > 0:
    print("STATUS: RECOMMEND CLEANUP | Executing this automated cleanup will reduce monthly infrastructure overhead.")
else:
    print("STATUS: OPTIMIZED | All active resources meet corporate utilization baselines.")
```
